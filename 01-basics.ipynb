## 1. Variables and core types

A **variable** is a name for a value. Python remembers the **type**.

| Type | Example | Chart field |
|------|---------|-------------|
| `str` (text) | `"Ana"` | name |
| `int` (whole number) | `15` | age in years |
| `float` (decimal) | `52.5` | weight in kg |
| `bool` (True/False) | `True` | has allergy |

**What you should see:** the four values and their types printed.

```python
# School clinic chart — change values and re-run
patient_name = "Ana"
age_years = 15
weight_kg = 52.5
has_allergy = True

print(patient_name, type(patient_name))
print(age_years, type(age_years))
print(weight_kg, type(weight_kg))
print(has_allergy, type(has_allergy))
```

## 2. Strings — clean a DNA sequence

Most lab text is messy: spaces, lowercase letters. You **strip**, **upper**, and **replace**.

**Biology:** DNA uses four bases: **A, T, G, C**.

**What you should see:** a clean uppercase sequence without spaces.

```python
raw_dna = "  a t g c a t  "
clean = raw_dna.strip().replace(" ", "").upper()
print("raw:", repr(raw_dna))
print("clean:", clean)
print("length:", len(clean))
print("first base:", clean[0])
print("last base:", clean[-1])
```

**List comprehension** — clean many pieces in one line.

`[part.strip() for part in text.split(",") if part.strip()]`

Read as: *for each part after splitting on comma, strip spaces; skip empty parts.*

```python
raw_codes = " ANA, bogdan , , Clara "
codes = [p.strip().title() for p in raw_codes.split(",") if p.strip()]
print(codes)
```

## 3. Lists — waiting room (order matters)

A **list** keeps order and can grow (`append`).

**What you should see:** four names, then a numbered waiting list.

```python
waiting = ["Ana", "Bogdan", "Clara"]
waiting.append("Dan")
print("waiting room:", waiting)

for i, name in enumerate(waiting, start=1):
    print(f"[{i}/{len(waiting)}] {name}")
```

## 4. Dictionaries — one patient chart (like a form)

A **dict** maps **key → value**. Nested dicts match a lab form.

**What you should see:** name and temperature from the chart.

```python
patient = {
    "name": "Ana",
    "age": 15,
    "weight_kg": 52.5,
    "vitals": {"temp_c": 37.8, "pulse": 88},
}

print(patient["name"])
print("temperature:", patient["vitals"]["temp_c"])
print("keys:", list(patient.keys()))
```

## 5. Tuples — blood pressure as a fixed pair

A **tuple** is ordered and **immutable** (you do not change it after creation). Unpacking: `sys, dia = bp`.

**What you should see:** `120 / 80`.

```python
bp = (120, 80)  # systolic, diastolic — practice numbers only
sys, dia = bp
print(f"{sys} / {dia}")
```

## 6. Sets — unique alleles and bad DNA letters

A **set** keeps **unique** values. Use it to dedupe or find unknowns.

**Biology:** alleles like `A` and `a`; DNA letters only `A T G C`.

**What you should see:** unique alleles, then invalid bases `{X, Z}`.

```python
alleles = ["A", "a", "A", "a", "A"]
print("unique alleles:", set(alleles))

seq = "ATGXCZ"
allowed = set("ATGC")
invalid = set(seq) - allowed
print("invalid bases:", invalid)
```

## 7. `if` / `elif` / `else` — fever label (exercise rule)

Indent with **4 spaces**. Combine with `and` / `or`.

These cutoffs are **for learning only**, not medical advice.

**What you should see:** one label for each temperature.

```python
def fever_label(temp_c):
    if temp_c >= 39.0:
        return "HIGH (practice rule)"
    elif temp_c >= 37.5:
        return "MILD (practice rule)"
    else:
        return "NORMAL (practice rule)"


for t in (36.8, 37.9, 39.2):
    print(t, "->", fever_label(t))
```

## 8. `for` loops — walk every chart

`for item in sequence:` is the default loop.

**What you should see:** who is ready for the nurse.

```python
charts = [
    {"name": "Ana", "ready": True},
    {"name": "Bogdan", "ready": False},
    {"name": "Clara", "ready": True},
]

for c in charts:
    if c["ready"]:
        print("call:", c["name"])
    else:
        print("wait:", c["name"])
```

## 9. `while` — re-check until stable

Use `while` when you do not know the count in advance (re-measure temperature).

**What you should see:** attempts until the practice temperature drops under 37.5.

```python
temps = [38.4, 38.1, 37.6, 37.2]  # pretend re-checks
i = 0
while i < len(temps) and temps[i] >= 37.5:
    print(f"check {i + 1}: {temps[i]} still warm (practice)")
    i += 1
if i < len(temps):
    print(f"check {i + 1}: {temps[i]} OK under practice cutoff")
else:
    print("still warm after all checks")
```

## 10. Capstone — parse a ward row → dict → label

Edit `ward_row`. Predict the output before you run.

Format: `name,weight_kg,temp_c`

```python
ward_row = "Ana,52.5,38.1"
name, weight_s, temp_s = [p.strip() for p in ward_row.split(",")]
chart = {
    "name": name,
    "weight_kg": float(weight_s),
    "temp_c": float(temp_s),
}

if chart["temp_c"] >= 37.5:
    label = "FEVER_FLAG (practice)"
else:
    label = "OK (practice)"

print(chart)
print(label)
```
