## 3. Regex — find start codon and temperature

**What you should see:** `ATG` match and `temp=38.2` value.

```python
import re

dna = "GGGATGCCCTTT"
m = re.search(r"ATG", dna)
print("start codon at index", m.start() if m else None)

line = "2026-03-28 wardA temp=38.2 pulse=90"
m2 = re.search(r"temp=([0-9.]+)", line)
print("temp", m2.group(1) if m2 else None)
```

## 4. Structured logging

**What you should see:** an INFO line about a practice event.

```python
import json
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s %(message)s")
log = logging.getLogger("clinic")


def log_json(event, **fields):
    payload = {"event": event, **fields}
    log.info(json.dumps(payload))


log_json("vitals_check", ward="A", patient="Ana", temp_c=37.8)
```

## 5. `subprocess` — safe local `date` only

Prefer a **list** of arguments (no `shell=True`).

**What you should see:** today's date string from the system.

```python
import subprocess

result = subprocess.run(["date"], capture_output=True, text=True, check=True)
print(result.stdout.strip())
print("Never pass untrusted strings to the shell.")
```

## 6. Capstone — write config + regex on a log line

**What you should see:** parsed ward and temperature from a saved snippet.

```python
from pathlib import Path
import re

p = Path("_clinic_notes") / "shift.log"
p.write_text("ward=A temp=38.1 pulse=92\n", encoding="utf-8")
line = p.read_text(encoding="utf-8").strip()
ward = re.search(r"ward=([A-Z])", line).group(1)
temp = float(re.search(r"temp=([0-9.]+)", line).group(1))
print({"ward": ward, "temp_c": temp})
```
