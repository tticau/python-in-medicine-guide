## 2. Practice dose for invented drug Alercin

Rule for this course only:

- `needed = weight_kg * mg_per_kg`
- if `needed > max_mg` → return `"OVER_MAX"` (do **not** give a number)
- else return the milligrams as a float

**What you should see:** a number for Ana, then `OVER_MAX` for a heavy practice case.

```python
def dose_mg(weight_kg, mg_per_kg=10, max_mg=1000):
    # Practice dose for invented Alercin — not a real medicine.
    needed = weight_kg * mg_per_kg
    if needed > max_mg:
        return "OVER_MAX"
    return needed


print("Ana:", dose_mg(52.5))          # 525 mg practice
print("heavy practice:", dose_mg(120))  # 1200 > 1000 -> OVER_MAX
```

## 3. Tablets and syrup form

- Tablets: how many whole tablets, and leftover mg.
- Syrup: millilitres from `mg / (mg_per_ml)`.

**What you should see:** tablet count + leftover, then mL rounded to 1 decimal.

```python
def tablets(dose, strength_mg=250):
    whole = int(dose // strength_mg)
    leftover = dose % strength_mg
    return whole, leftover


def syrup_ml(dose, mg_per_ml=25):
    return round(dose / mg_per_ml, 1)


d = 375  # practice mg
print("tablets:", tablets(d))
print("syrup mL:", syrup_ml(d))
```

## 4. Chemistry — moles and dilution

School formulas:

- moles: `n = mass_g / molar_mass`
- dilution: `C1 * V1 = C2 * V2` → `V1 = (C2 * V2) / C1`

Example: saline / glucose numbers for practice only.

**What you should see:** moles of glucose sample, then V1 for a dilution.

```python
def moles(mass_g, molar_mass):
    return mass_g / molar_mass


def dilution_v1(c1, c2, v2):
    # Volume of stock needed. All units must match (e.g. mol/L and L).
    return (c2 * v2) / c1


# Glucose C6H12O6 molar mass ≈ 180 g/mol (school value)
print("moles:", round(moles(9.0, 180.0), 4))
print("V1 liters:", dilution_v1(c1=1.0, c2=0.1, v2=0.5))
```

## 5. Count atoms in a simple formula string

We only handle patterns like `C6H12O6` (element + optional number).

**What you should see:** `{'C': 6, 'H': 12, 'O': 6}`.

```python
import re


def atom_counts(formula):
    parts = re.findall(r"([A-Z][a-z]?)(\d*)", formula)
    counts = {}
    for element, num in parts:
        n = int(num) if num else 1
        counts[element] = counts.get(element, 0) + n
    return counts


print(atom_counts("C6H12O6"))
```

## 6. DNA complement and RNA

DNA pairing: A↔T, C↔G.  
For a practice **RNA** transcript from a template DNA strand, use A→U, T→A, C→G, G→C.

**What you should see:** complement `TACG`, RNA `UACG`.

```python
PAIRS = {"A": "T", "T": "A", "C": "G", "G": "C"}
RNA_FROM_TEMPLATE = {"A": "U", "T": "A", "C": "G", "G": "C"}


def complement(dna):
    dna = dna.upper().replace(" ", "")
    return "".join(PAIRS[b] for b in dna)


def to_rna(template_dna):
    dna = template_dna.upper().replace(" ", "")
    return "".join(RNA_FROM_TEMPLATE[b] for b in dna)


print(complement("ATGC"))
print(to_rna("ATGC"))
```

## 7. Punnett square — monohybrid `Aa` × `Aa`

Each parent gives one allele. Four combinations. Expected school ratio: **1 AA : 2 Aa : 1 aa**.

**What you should see:** counts of AA, Aa, aa.

```python
def punnett_counts(parent1, parent2):
    # parent1/parent2 are 2-letter genotypes like 'Aa'
    counts = {}
    for a in parent1:
        for b in parent2:
            geno = "".join(sorted([a, b], key=lambda ch: (ch.islower(), ch)))
            if geno == "aA":
                geno = "Aa"
            counts[geno] = counts.get(geno, 0) + 1
    return counts


print(punnett_counts("Aa", "Aa"))
```

## 10. Classes — Patient and Medicine

## 11. Inheritance — Tablet vs Syrup

Shared behaviour in `Medicine`; each form explains **how_to_give**.

**What you should see:** two different instruction strings.

```python
class Medicine:
    def __init__(self, name, mg_per_kg, max_mg):
        self.name = name
        self.mg_per_kg = mg_per_kg
        self.max_mg = max_mg

    def practice_dose(self, weight_kg):
        needed = weight_kg * self.mg_per_kg
        if needed > self.max_mg:
            return "OVER_MAX"
        return needed


class Tablet(Medicine):
    def how_to_give(self, dose):
        if dose == "OVER_MAX":
            return "do not give (practice OVER_MAX)"
        return f"swallow tablets for {dose} mg of {self.name} (practice)"


class Syrup(Medicine):
    def __init__(self, name, mg_per_kg, max_mg, mg_per_ml):
        super().__init__(name, mg_per_kg, max_mg)
        self.mg_per_ml = mg_per_ml

    def how_to_give(self, dose):
        if dose == "OVER_MAX":
            return "do not give (practice OVER_MAX)"
        ml = round(dose / self.mg_per_ml, 1)
        return f"measure {ml} mL of {self.name} syrup (practice)"


alercin_tab = Tablet("Alercin", 10, 1000)
alercin_syr = Syrup("Alercin", 10, 1000, 25)
d = alercin_tab.practice_dose(52.5)
print(alercin_tab.how_to_give(d))
print(alercin_syr.how_to_give(d))
```

```python
def bmi(weight_kg, height_m):
    return round(weight_kg / height_m ** 2, 1)


print(bmi(52.5, 1.60))
```

```python
def gc_content(dna):
    dna = dna.upper().replace(" ", "")
    if not dna:
        return 0.0
    gc = sum(1 for b in dna if b in "GC")
    return round(100 * gc / len(dna), 1)


print(gc_content("ATGCATGC"))
```
