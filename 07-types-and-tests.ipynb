## 1. Why type hints?

**What you should see:** a capped practice dose as `float | str` style result documented in the signature (we use simple hints).

```python
def dose_mg(weight_kg: float, mg_per_kg: float = 10.0, max_mg: float = 1000.0):
    needed = weight_kg * mg_per_kg
    if needed > max_mg:
        return "OVER_MAX"
    return needed


print(dose_mg(52.5))
```

## 2. `Optional` — temperature may be missing

**What you should see:** `None` handled.

```python
from typing import Optional


def label_temp(temp_c: Optional[float]) -> str:
    if temp_c is None:
        return "MISSING"
    if temp_c >= 37.5:
        return "FEVER_FLAG (practice)"
    return "OK (practice)"


print(label_temp(None))
print(label_temp(36.9))
```

## 3. `TypedDict` — lab result shape

**What you should see:** a formatted lab line.

```python
from typing import TypedDict


class LabResult(TypedDict):
    name: str
    value: float
    unit: str


def as_line(m: LabResult) -> str:
    return f"{m['name']}={m['value']}{m['unit']}"


print(as_line({"name": "glucose", "value": 5.2, "unit": "mmol/L"}))
```

## 4. Pytest style — plain `assert`

In a real file you run `pytest`. Here we run asserts inline.

**What you should see:** no error (all asserts pass).

```python
def complement(dna: str) -> str:
    pairs = {"A": "T", "T": "A", "C": "G", "G": "C"}
    return "".join(pairs[b] for b in dna.upper())


assert complement("ATGC") == "TACG"
assert dose_mg(10, 10, 1000) == 100
assert dose_mg(120, 10, 1000) == "OVER_MAX"
print("all asserts passed")
```

## 5. Table of cases (parametrize idea)

**What you should see:** each case OK.

```python
def classify(code: int) -> str:
    if 200 <= code <= 299:
        return "ok"
    if 400 <= code <= 499:
        return "client"
    if 500 <= code <= 599:
        return "server"
    return "other"


cases = [(200, "ok"), (404, "client"), (503, "server")]
for code, expected in cases:
    assert classify(code) == expected
    print(code, "OK")
```

## 6. Punnett ratio test

**What you should see:** 1:2:1 for `Aa` x `Aa`.

```python
def punnett_counts(p1: str, p2: str) -> dict:
    counts: dict[str, int] = {}
    for a in p1:
        for b in p2:
            geno = "".join(sorted([a, b], key=lambda ch: (ch.islower(), ch)))
            if geno == "aA":
                geno = "Aa"
            counts[geno] = counts.get(geno, 0) + 1
    return counts


c = punnett_counts("Aa", "Aa")
assert c.get("AA") == 1
assert c.get("Aa") == 2
assert c.get("aa") == 1
print(c)
```

## 7. Capstone — typed helpers + tests together

**What you should see:** `tests passed`.

```python
from typing import Optional, TypedDict, Union


class PatientChart(TypedDict):
    name: str
    weight_kg: float
    temp_c: Optional[float]


def practice_dose(weight_kg: float, mg_per_kg: float = 10.0, max_mg: float = 1000.0) -> Union[float, str]:
    needed = weight_kg * mg_per_kg
    if needed > max_mg:
        return "OVER_MAX"
    return needed


def fever_flag(chart: PatientChart) -> str:
    t = chart["temp_c"]
    if t is None:
        return "MISSING"
    return "FEVER_FLAG (practice)" if t >= 37.5 else "OK (practice)"


assert practice_dose(50) == 500
assert practice_dose(120) == "OVER_MAX"
assert fever_flag({"name": "Ana", "weight_kg": 52.5, "temp_c": None}) == "MISSING"
print("tests passed")
```
